```python
import numpy as np
import pandas as pd


# TASK 1: Show variance calculation step by step
steps = np.array([4500, 7000, 5000, 8000, 4000, 9000, 6000])

mean_steps = np.mean(steps)
differences = steps - mean_steps
squared_differences = differences ** 2

calculation = pd.DataFrame({
    "Steps": steps,
    "Difference from mean": differences,
    "Squared difference": squared_differences
})

# Calculate using the formula, rather than np.var().
variance_steps = np.sum(squared_differences) / len(steps)

print("TASK 1: Daily Steps Variance")
print(f"Total steps: {np.sum(steps)}")
print(f"Mean: {mean_steps:.2f}")
print(calculation.round(2).to_string(index=False))
print(f"Sum of squared differences: {np.sum(squared_differences):.2f}")
print(f"Population variance: {variance_steps:.2f} steps squared")


# TASK 2: Function to calculate standard deviation
def calculate_standard_deviation(scores):
    """Return population standard deviation rounded to 2 decimals."""
    if len(scores) == 0:
        raise ValueError("Please provide at least one score.")

    return round(float(np.std(scores, ddof=0)), 2)


# Example exam scores
scores = [70, 80, 90, 100]

print("\nTASK 2: Exam Score Standard Deviation")
print(f"Standard deviation: {calculate_standard_deviation(scores):.2f}")


# TASK 3: Compare spending consistency
friend_a = [200, 200, 200, 200, 200]
friend_b = [100, 300, 150, 400, 50]

spending_summary = pd.DataFrame({
    "Friend": ["A", "B"],
    "Mean spending": [np.mean(friend_a), np.mean(friend_b)],
    "Standard deviation": [
        calculate_standard_deviation(friend_a),
        calculate_standard_deviation(friend_b)
    ]
})

print("\nTASK 3: Spending Comparison")
print(spending_summary.to_string(index=False))
print(
    "\nFriend A has more consistent spending because the amount "
    "is Rs. 200 every day, giving a standard deviation of zero. "
    "Friend B has the same average spending, but the daily amounts "
    "vary, giving a standard deviation of Rs. 130.38."
)


# TASK 4: Salary variance and standard deviation
salaries = np.array([28000, 29000, 27000, 60000, 26500, 27500])

salary_mean = np.mean(salaries)
salary_variance = np.var(salaries, ddof=0)
salary_std = np.std(salaries, ddof=0)

print("\nTASK 4: Salary Analysis")
print(f"Mean salary: Rs. {salary_mean:,.2f}")
print(f"Population variance: {salary_variance:,.2f} rupees squared")
print(f"Population standard deviation: Rs. {salary_std:,.2f}")

print(
    "\nFive salaries are between Rs. 26,500 and Rs. 29,000, "
    "while one is Rs. 60,000. This higher salary increases "
    "the spread and raises the mean above the other five salaries."
)
```

```python
import numpy as np
from scipy import stats

alpha = 0.05

# TASK 1: Instagram vs YouTube hypotheses
print("TASK 1: Null and Alternative Hypotheses")
print("H0: Mean daily Instagram time <= mean daily YouTube time.")
print("H1: Mean daily Instagram time > mean daily YouTube time.")
print("This is a right-tailed test.")


# TASK 2: Two-sample t-test for Zomato order values
mean_a, std_a, n_a = 350, 50, 30
mean_b, std_b, n_b = 320, 50, 30

# Treat the given standard deviations as sample standard deviations.
# Use a two-sided Welch independent-samples t-test.
t_statistic, p_value = stats.ttest_ind_from_stats(
    mean1=mean_a,
    std1=std_a,
    nobs1=n_a,
    mean2=mean_b,
    std2=std_b,
    nobs2=n_b,
    equal_var=False,
    alternative="two-sided"
)

print("\nTASK 2: Zomato Two-Sample T-Test")
print("H0: The two home pages have equal population mean order values.")
print("H1: The population mean order values differ.")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Significance level: {alpha}")

if p_value < alpha:
    print("Decision: Reject H0.")
    print(
        "There is statistically significant evidence of a difference "
        "in mean order values. Group A has the higher observed mean."
    )
else:
    print("Decision: Fail to reject H0.")
    print("There is insufficient evidence of a difference.")


# TASK 3: Design an A/B test
print("\nTASK 3: A/B Test for a Paytm Payment Button")
print("Feature: A redesigned payment button.")
print("Control group: Users shown the existing button.")
print("Variant group: Users shown the redesigned button.")
print(
    "Assignment: Randomly assign eligible users to two groups, "
    "keeping each user in the same group throughout the experiment."
)
print(
    "Metric: Payment completion rate = users completing a payment "
    "/ all assigned eligible users."
)
print("H0: New-button completion rate <= old-button completion rate.")
print("H1: New-button completion rate > old-button completion rate.")
print(
    "Choose the sample size and test duration before starting. "
    "Run both groups simultaneously and compare completion rates "
    "using a one-sided two-proportion test."
)


# TASK 4: Interpret the Flipkart click counts
print("\nTASK 4: Flipkart Wishlist Button")
print(
    "The question gives 520 new-button clicks and 480 old-button "
    "clicks, but does not clearly give the number exposed to each "
    "button. Click counts alone are insufficient to compare rates."
)

# Illustrative assumption: 1,000 independent users PER group.
# This means 2,000 users overall, not 1,000 overall.
clicks_new, users_new = 520, 1000
clicks_old, users_old = 480, 1000

rate_new = clicks_new / users_new
rate_old = clicks_old / users_old

# Under H0, estimate a shared click probability.
pooled_rate = (
    (clicks_new + clicks_old) / (users_new + users_old)
)

standard_error = np.sqrt(
    pooled_rate * (1 - pooled_rate)
    * (1 / users_new + 1 / users_old)
)

z_statistic = (rate_new - rate_old) / standard_error
click_p_value = 2 * stats.norm.sf(abs(z_statistic))

print("\nIllustration assuming 1,000 users PER group:")
print(f"New-button click rate: {rate_new:.1%}")
print(f"Old-button click rate: {rate_old:.1%}")
print(f"Two-sided p-value: {click_p_value:.4f}")

if click_p_value < alpha:
    print("Reject H0: Evidence of different click rates.")
else:
    print("Fail to reject H0: Insufficient evidence of different click rates.")

print(
    "\nA p-value measures how unusual a result at least this extreme "
    "would be if H0 were true, under the test assumptions. "
    "It is not the probability that H0 is true."
)
```

```python
import random


# TASK 1: Calculate the probability of an event
def calculate_probability(event_count, total_outcomes):
    """Calculate probability for equally likely outcomes or user counts."""
    if total_outcomes <= 0:
        raise ValueError("Total outcomes must be greater than zero.")

    if not 0 <= event_count <= total_outcomes:
        raise ValueError("Event count must be between 0 and total outcomes.")

    return event_count / total_outcomes


# A fair coin has two equally likely outcomes: Heads and Tails.
heads_probability = calculate_probability(1, 2)

print("TASK 1: Fair Coin")
print(f"Probability of Heads: {heads_probability:.2f}")
print(f"Percentage: {heads_probability:.0%}")


# TASK 2: Simulate rolling a fair six-sided die 100 times
rng = random.Random(42)
six_count = 0
number_of_rolls = 100

for _ in range(number_of_rolls):
    roll = rng.randint(1, 6)

    if roll == 6:
        six_count += 1

experimental_probability = calculate_probability(
    six_count, number_of_rolls
)

print("\nTASK 2: Dice Simulation")
print("Number of rolls:", number_of_rolls)
print("Number of sixes:", six_count)
print(f"Experimental probability: {experimental_probability:.3f}")
print(f"Theoretical probability: {1 / 6:.3f}")
print(
    "The experimental probability may differ from 1/6 "
    "because the simulation uses only 100 random rolls."
)


# TASK 3: Probability of selecting a Pop song
# Example history containing exactly 20 song entries.
genres = [
    "pop", "rock", "pop", "hiphop", "jazz",
    "pop", "classical", "rock", "pop", "hiphop",
    "pop", "jazz", "rock", "pop", "classical",
    "pop", "hiphop", "rock", "pop", "jazz"
]

pop_count = genres.count("pop")
pop_probability = calculate_probability(pop_count, len(genres))

print("\nTASK 3: Spotify Song History")
print("Total songs:", len(genres))
print("Pop songs:", pop_count)
print(f"Probability of selecting Pop: {pop_probability:.2f}")
print(f"Percentage: {pop_probability:.0%}")


# TASK 4: Probability a user ordered both food and dessert
def probability_both(food_users, dessert_users):
    # Sets remove duplicate user IDs.
    food_set = set(food_users)
    dessert_set = set(dessert_users)

    both_users = food_set & dessert_set
    total_unique_users = food_set | dessert_set

    return calculate_probability(
        len(both_users), len(total_unique_users)
    )


food_users = ["user1", "user2", "user3", "user4", "user5"]
dessert_users = ["user3", "user4", "user5", "user6"]

both_probability = probability_both(food_users, dessert_users)

print("\nTASK 4: Food and Dessert")
print("Users who ordered both:", sorted(set(food_users) & set(dessert_users)))
print("Total unique users:", len(set(food_users) | set(dessert_users)))
print(f"Probability of both: {both_probability:.2f}")
print(
    "Here, a user is selected from everyone who ordered "
    "food, dessert, or both."
)


# TASK 5: Conditional probability for recommendations
# Illustrative purchase data, not actual Flipkart records.
phone_buyers = {"user1", "user2", "user3", "user4", "user5"}
headphone_buyers = {"user2", "user4", "user5", "user6"}

bought_both = phone_buyers & headphone_buyers

conditional_probability = calculate_probability(
    len(bought_both), len(phone_buyers)
)

print("\nTASK 5: Conditional Probability")
print("Phone buyers:", len(phone_buyers))
print("Buyers of both products:", len(bought_both))
print(
    "P(Headphones | Phone) = "
    f"{conditional_probability:.2f} "
    f"({conditional_probability:.0%})"
)
print(
    "In this example, 60% of phone buyers also bought headphones. "
    "An app could use this pattern as one signal for recommending "
    "headphones to phone buyers, alongside other user preferences."
)
```

```python
import pandas as pd
import re


# TASK 1: Function for Bayes' theorem
def bayes_posterior(prior, likelihood, evidence):
    """
    prior      = P(A)
    likelihood = P(B | A)
    evidence   = P(B)
    returns      P(A | B)
    """
    if not 0 <= prior <= 1:
        raise ValueError("Prior must be between 0 and 1.")

    if not 0 <= likelihood <= 1:
        raise ValueError("Likelihood must be between 0 and 1.")

    if not 0 < evidence <= 1:
        raise ValueError("Evidence must be greater than 0 and at most 1.")

    joint_probability = prior * likelihood

    if joint_probability > evidence:
        raise ValueError("The supplied probabilities are inconsistent.")

    return joint_probability / evidence


print("TASK 1:")
example = bayes_posterior(
    prior=0.10,
    likelihood=0.80,
    evidence=0.26
)
print(f"Example posterior probability: {example:.2%}")


# TASK 2: Premium Swiggy restaurant given a discount
# Assume every restaurant is either premium or regular.

prior_premium = 0.10                  # P(Premium)
prior_regular = 1 - prior_premium     # P(Regular)

discount_given_premium = 0.80         # P(Discount | Premium)
discount_given_regular = 0.20         # P(Discount | Regular)

# Evidence: P(Discount), using the law of total probability
discount_probability = (
    prior_premium * discount_given_premium
    + prior_regular * discount_given_regular
)

premium_given_discount = bayes_posterior(
    prior=prior_premium,
    likelihood=discount_given_premium,
    evidence=discount_probability
)

print("\nTASK 2: Swiggy")
print(f"P(Discount): {discount_probability:.2%}")
print(f"P(Premium | Discount): {premium_given_discount:.2%}")
print(
    "Among restaurants offering a discount, "
    "approximately 30.77% are premium under these assumptions."
)


# TASK 3: Basic spam detection
# Synthetic, labeled dataset of exactly 20 email subjects.
emails = pd.DataFrame({
    "Subject": [
        "Claim your free prize",
        "Free cash reward",
        "Get a free vacation",
        "Your free gift awaits",
        "Win free tickets",
        "Free bonus offer",
        "Collect your reward now",
        "Exclusive lottery winner",
        "Urgent cash offer",
        "Huge discount today",
        "Free workshop registration",
        "Are you free for lunch?",
        "Team meeting tomorrow",
        "Project status update",
        "Your invoice is attached",
        "Dinner plans tonight",
        "Appointment confirmation",
        "Monthly account statement",
        "Assignment submission reminder",
        "Photos from our trip"
    ],
    "Is Spam": [True] * 10 + [False] * 10
})

# Match the whole word "free", ignoring capitalization.
emails["Contains Free"] = emails["Subject"].apply(
    lambda subject: bool(re.search(r"\bfree\b", subject, re.IGNORECASE))
)

spam_mask = emails["Is Spam"]

# Estimate probabilities from the labeled example data.
prior_spam = spam_mask.mean()

free_given_spam = emails.loc[
    spam_mask, "Contains Free"
].mean()

free_given_not_spam = emails.loc[
    ~spam_mask, "Contains Free"
].mean()

evidence_free = (
    prior_spam * free_given_spam
    + (1 - prior_spam) * free_given_not_spam
)

spam_given_free = bayes_posterior(
    prior=prior_spam,
    likelihood=free_given_spam,
    evidence=evidence_free
)

print("\nTASK 3: Spam Detection")
print(emails.to_string(index=False))
print(f"\nP(Spam): {prior_spam:.2%}")
print(f"P(Free | Spam): {free_given_spam:.2%}")
print(f"P(Free | Not Spam): {free_given_not_spam:.2%}")
print(f"P(Free): {evidence_free:.2%}")
print(f"P(Spam | Free): {spam_given_free:.2%}")
print(
    "In this synthetic dataset, 75% of emails containing 'free' "
    "are spam. This is a demonstration, not a validated spam filter."
)


# TASK 4: AI-generated app example and its solution
prompt = (
    "Generate a hypothetical Bayes theorem example about Flipkart "
    "using membership status and express delivery. Include the "
    "prior and conditional probabilities needed to solve it."
)

scenario = (
    "Suppose 30% of Flipkart customers have a membership. "
    "Of members, 60% choose express delivery; "
    "of non-members, 20% choose express delivery. "
    "If a customer chooses express delivery, what is the "
    "probability that they are a member?"
)

# Hypothetical probabilities, not actual Flipkart statistics.
prior_member = 0.30
express_given_member = 0.60
express_given_non_member = 0.20

evidence_express = (
    prior_member * express_given_member
    + (1 - prior_member) * express_given_non_member
)

member_given_express = bayes_posterior(
    prior=prior_member,
    likelihood=express_given_member,
    evidence=evidence_express
)

print("\nTASK 4: AI-Generated Example")
print("Prompt:", prompt)
print("\nScenario:", scenario)
print(f"\nP(Express): {evidence_express:.2%}")
print(f"P(Member | Express): {member_given_express:.2%}")
```
